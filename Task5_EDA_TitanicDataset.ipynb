```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

sns.set(style="darkgrid")
```

```python
df = pd.read_csv('Titanic-Dataset.csv')
```

```python
df.head()
df.info()
df.describe(include='all')
df.isnull().sum()
df.shape
```

```python
sns.histplot(df['Age'].dropna(), kde=True)
plt.title('Distribution of Age')
plt.show()
```

```python
sns.boxplot(x='Age', data=df)
plt.title('Boxplot of Age')
plt.show()
```

```python
sns.countplot(x='Sex', hue='Survived', data=df)
plt.title('Survival by Sex')
plt.show()
```

```python
sns.scatterplot(x='Age', y='Fare', hue='Survived', data=df)
plt.title('Age vs Fare colored by Survival')
plt.show()
```

```python
sns.pairplot(df[['Survived', 'Age', 'Fare', 'Pclass', 'SibSp', 'Parch']].dropna().astype(float), hue='Survived')
plt.suptitle('Pairplot of Numerical Features', y=1.02)
plt.show()
```

```python
sns.heatmap(df.corr(numeric_only=True), annot=True, cmap='coolwarm')
plt.title('Correlation Heatmap')
plt.show()
```

 Summary of Insights

- Most passengers were between 20-40 years old.
- More women survived than men.
- First-class passengers had higher survival rates.
- Passengers who paid more had higher survival chances.
- Some missing values are present in Age and Cabin.
